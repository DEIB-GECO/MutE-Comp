```python
import configparser
import os

import pandas as pd
```

```python
os.chdir('../')
```

```python
dataset = 'all_noHypoHyper'
```

```python
data = configparser.ConfigParser(interpolation=configparser.ExtendedInterpolation())
data.read('../data.ini')
data = data['DEFAULT']
```

```python
dataMutations = pd.read_csv('../' + data[f'{dataset}___mutations'], sep='\t', comment='#', engine='python')
dataMutations_RAS = pd.read_csv('../' + data[f'{dataset}_RAS___mutations'], sep='\t', comment='#', engine='python')
dataMutations_nonRAS = pd.read_csv('../' + data[f'{dataset}_nonRAS___mutations'], sep='\t', comment='#', engine='python')
```

```python
dataMutations.groupby('Hugo_Symbol').filter(lambda x: x['Protein_Length'].nunique() > 1)[['Hugo_Symbol', 'Transcript_ID', 'Gene_Length', 'Protein_Length']].drop_duplicates()
# there are equals symbol with different transcript lengths
# TODO: use Transcript_ID as symbol, correct solution?
```

```python
genesLengths = dataMutations[['Transcript_ID', 'Protein_Length', 'Hugo_Symbol']]
genesLengths = genesLengths.rename(columns={'Transcript_ID': 'symbol', 'Protein_Length': 'length', 'Hugo_Symbol': 'gene'})
genesLengths = genesLengths.drop_duplicates()

genesLengths_RAS = dataMutations_RAS[['Transcript_ID', 'Protein_Length', 'Hugo_Symbol']]
genesLengths_RAS = genesLengths_RAS.rename(columns={'Transcript_ID': 'symbol', 'Protein_Length': 'length', 'Hugo_Symbol': 'gene'})
genesLengths_RAS = genesLengths_RAS.drop_duplicates()

genesLengths_nonRAS = dataMutations_nonRAS[['Transcript_ID', 'Protein_Length', 'Hugo_Symbol']]
genesLengths_nonRAS = genesLengths_nonRAS.rename(columns={'Transcript_ID': 'symbol', 'Protein_Length': 'length', 'Hugo_Symbol': 'gene'})
genesLengths_nonRAS = genesLengths_nonRAS.drop_duplicates()
```

```python
genesLengths
```

```python
mutationPositions = dataMutations[['Transcript_ID', 'Protein_position', 'Hugo_Symbol']]
mutationPositions = mutationPositions.rename(columns={'Transcript_ID': 'symbol', 'Protein_position': 'position', 'Hugo_Symbol': 'gene'})

mutationPositions_RAS = dataMutations_RAS[['Transcript_ID', 'Protein_position', 'Hugo_Symbol']]
mutationPositions_RAS = mutationPositions_RAS.rename(columns={'Transcript_ID': 'symbol', 'Protein_position': 'position', 'Hugo_Symbol': 'gene'})

mutationPositions_nonRAS = dataMutations_nonRAS[['Transcript_ID', 'Protein_position', 'Hugo_Symbol']]
mutationPositions_nonRAS = mutationPositions_nonRAS.rename(columns={'Transcript_ID': 'symbol', 'Protein_position': 'position', 'Hugo_Symbol': 'gene'})
```

```python
mutationPositions
```

```python
genesLengths.to_csv('../' + data['mutclustsw___all_noHypoHyper___in_dir'] + 'genesLengths.txt', sep='\t', index=False)
genesLengths_RAS.to_csv('../' + data['mutclustsw___all_noHypoHyper_RAS___in_dir'] + 'genesLengths.txt', sep='\t', index=False)
genesLengths_nonRAS.to_csv('../' + data['mutclustsw___all_noHypoHyper_nonRAS___in_dir'] + 'genesLengths.txt', sep='\t', index=False)
```

```python
mutationPositions.to_csv('../' + data['mutclustsw___all_noHypoHyper___in_dir'] + 'mutationPositions.txt', sep='\t', index=False)
mutationPositions_RAS.to_csv('../' + data['mutclustsw___all_noHypoHyper_RAS___in_dir'] + 'mutationPositions.txt', sep='\t', index=False)
mutationPositions_nonRAS.to_csv('../' + data['mutclustsw___all_noHypoHyper_nonRAS___in_dir'] + 'mutationPositions.txt', sep='\t', index=False)
```
