## Score analysis

```python
score_qfreq = 0.01
```

```python
numGenes_pvLower_max_HN = pd.DataFrame(
    [(i, len(muffinMax_HN.loc[muffinMax_HN['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_RAS_max_HN = pd.DataFrame(
    [(i, len(muffinMax_HN_RAS.loc[muffinMax_HN_RAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_nonRAS_max_HN = pd.DataFrame(
    [(i, len(muffinMax_HN_nonRAS.loc[muffinMax_HN_nonRAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)

numGenes_pvLower_max_S = pd.DataFrame(
    [(i, len(muffinMax_S.loc[muffinMax_S['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_RAS_max_S = pd.DataFrame(
    [(i, len(muffinMax_S_RAS.loc[muffinMax_S_RAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_nonRAS_max_S = pd.DataFrame(
    [(i, len(muffinMax_S_nonRAS.loc[muffinMax_S_nonRAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_sum_HN = pd.DataFrame(
    [(i, len(muffinSum_HN.loc[muffinSum_HN['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_RAS_sum_HN = pd.DataFrame(
    [(i, len(muffinSum_HN_RAS.loc[muffinSum_HN_RAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_nonRAS_sum_HN = pd.DataFrame(
    [(i, len(muffinSum_HN_nonRAS.loc[muffinSum_HN_nonRAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)

numGenes_pvLower_sum_S = pd.DataFrame(
    [(i, len(muffinSum_S.loc[muffinSum_S['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_RAS_sum_S = pd.DataFrame(
    [(i, len(muffinSum_S_RAS.loc[muffinSum_S_RAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
numGenes_pvLower_nonRAS_sum_S = pd.DataFrame(
    [(i, len(muffinSum_S_nonRAS.loc[muffinSum_S_nonRAS['Score'] >= i, 'Gene_Symbol'])) for i in np.clip(np.arange(score_qfreq, 1+score_qfreq, score_qfreq), a_min=0, a_max=1)],
    columns=['Score', 'numGenes']
)
```

```python
# plot
plt.figure(figsize=(15, 5))
plt.title(dataset + ' - Cumulative number of significant genes')

plt.plot(numGenes_pvLower_max_HN.set_index('Score'), label='all - max - HumanNet', color='blue')
# plt.plot(numGenes_pvLower_RAS_max_HN.set_index('Score'), label='RAS - max - HumanNet', color='red')
# plt.plot(numGenes_pvLower_nonRAS_max_HN.set_index('Score'), label='nonRAS - max - HumanNet', color='green')

plt.plot(numGenes_pvLower_max_S.set_index('Score'), label='all - max - STRINGv10', color='lightblue')
# plt.plot(numGenes_pvLower_RAS_max_S.set_index('Score'), label='RAS - max - STRINGv10', color='orange')
# plt.plot(numGenes_pvLower_nonRAS_max_S.set_index('Score'), label='nonRAS - max - STRINGv10', color='lightgreen')

plt.plot(numGenes_pvLower_sum_HN.set_index('Score'), label='all - sum - HumanNet', color='blue', linestyle='dashed')
# plt.plot(numGenes_pvLower_RAS_sum_HN.set_index('Score'), label='RAS - sum - HumanNet', color='red', linestyle='dashed')
# plt.plot(numGenes_pvLower_nonRAS_sum_HN.set_index('Score'), label='nonRAS - sum - HumanNet', color='green', linestyle='dashed')

plt.plot(numGenes_pvLower_sum_S.set_index('Score'), label='all - sum - STRINGv10', color='lightblue', linestyle='dashed')
# plt.plot(numGenes_pvLower_RAS_sum_S.set_index('Score'), label='RAS - sum - STRINGv10', color='orange', linestyle='dashed')
# plt.plot(numGenes_pvLower_nonRAS_sum_S.set_index('Score'), label='nonRAS - sum - STRINGv10', color='lightgreen', linestyle='dashed')

plt.xlabel('MUFFINN-Score', fontsize=11)
plt.ylabel('Number of significant genes', fontsize=11)
plt.legend(loc='upper right', fontsize=10)
plt.savefig("../" + data[f'plots___{dataset}'] + dataset + '_MUFFINN_cumulativeNumSignificantGenes.png')
plt.show()
```
