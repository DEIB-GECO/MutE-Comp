```python
import configparser
import os

import pandas as pd
```

```python
os.chdir('../')
```

```python
dataset = 'all_noHypoHyper'
```

```python
data = configparser.ConfigParser(interpolation=configparser.ExtendedInterpolation())
data.read('../data.ini')
data = data['DEFAULT']
```

```python
mutations = pd.read_csv("../" + data[f'{dataset}___mutations'], sep="\t", comment='#')
mutations_RAS = pd.read_csv("../" + data[f'{dataset}_RAS___mutations'], sep="\t", comment='#')
mutations_nonRAS = pd.read_csv("../" + data[f'{dataset}_nonRAS___mutations'], sep="\t", comment='#')
```

```python
mutations
```

```python
occurrences = mutations['Hugo_Symbol'].value_counts().reset_index()
occurrences_RAS = mutations_RAS['Hugo_Symbol'].value_counts().reset_index()
occurrences_nonRAS = mutations_nonRAS['Hugo_Symbol'].value_counts().reset_index()
```

```python
occurrences
```

```python
hugo_to_entrez = pd.read_csv("../" + data['muffinn'] + '/hugo_symbol_to_entrez_id.csv', sep="\t")
hugo_to_entrez
```

```python
occurrences_entrez = occurrences.merge(hugo_to_entrez, left_on='Hugo_Symbol', right_on='hugo_symbol', how='left').rename(columns={'count': 'occurrences'})[['hugo_symbol', 'entrez_id', 'occurrences']]
occurrences_entrez_RAS = occurrences_RAS.merge(hugo_to_entrez, left_on='Hugo_Symbol', right_on='hugo_symbol', how='left').rename(columns={'count': 'occurrences'})[['hugo_symbol', 'entrez_id', 'occurrences']]
occurrences_entrez_nonRAS = occurrences_nonRAS.merge(hugo_to_entrez, left_on='Hugo_Symbol', right_on='hugo_symbol', how='left').rename(columns={'count': 'occurrences'})[['hugo_symbol', 'entrez_id', 'occurrences']]
```

```python
occurrences_entrez
```

```python
occurrences_entrez[occurrences_entrez['entrez_id'].isna()]
```

```python
# TODO: remove na??
occurrences_entrez = occurrences_entrez[occurrences_entrez['entrez_id'].notna()]
occurrences_entrez_RAS = occurrences_entrez_RAS[occurrences_entrez_RAS['entrez_id'].notna()]
occurrences_entrez_nonRAS = occurrences_entrez_nonRAS[occurrences_entrez_nonRAS['entrez_id'].notna()]
```

```python
occurrences_entrez[occurrences_entrez['entrez_id'].str.contains('|', regex=False)]
```

```python
# TODO: take first??
occurrences_entrez.loc[:,'entrez_id'] = occurrences_entrez['entrez_id'].apply(lambda x: x.split('|')[0])
occurrences_entrez_RAS.loc[:,'entrez_id'] = occurrences_entrez_RAS['entrez_id'].apply(lambda x: x.split('|')[0])
occurrences_entrez_nonRAS.loc[:,'entrez_id'] = occurrences_entrez_nonRAS['entrez_id'].apply(lambda x: x.split('|')[0])
```

```python
occurrences_entrez = occurrences_entrez[['entrez_id', 'occurrences']]
occurrences_entrez_RAS = occurrences_entrez_RAS[['entrez_id', 'occurrences']]
occurrences_entrez_nonRAS = occurrences_entrez_nonRAS[['entrez_id', 'occurrences']]
```

```python
occurrences_entrez
```

```python
occurrences_entrez.to_csv("../" + data[f'muffinn___{dataset}___in'], sep="\t", index=False, header=False)
occurrences_entrez_RAS.to_csv("../" + data[f'muffinn___{dataset}_RAS___in'], sep="\t", index=False, header=False)
occurrences_entrez_nonRAS.to_csv("../" + data[f'muffinn___{dataset}_nonRAS___in'], sep="\t", index=False, header=False)
```
