```python
import configparser
import os

import pandas as pd
```

```python
os.chdir('../')
```

```python
dataset = 'all_noHypoHyper'
```

```python
data = configparser.ConfigParser(interpolation=configparser.ExtendedInterpolation())
data.read('../data.ini')
data = data['DEFAULT']
```

```python
mutsig = pd.read_csv("../" + data[f'mutsig___{dataset}___out_dir'] + '/sig_genes.txt', sep='\t', engine='python', comment='#')
mutsig_RAS = pd.read_csv("../" + data[f'mutsig___{dataset}_RAS___out_dir'] + '/sig_genes.txt', sep='\t', engine='python', comment='#')
mutsig_nonRAS = pd.read_csv("../" + data[f'mutsig___{dataset}_nonRAS___out_dir'] + '/sig_genes.txt', sep='\t', engine='python', comment='#')
```

```python
mutsig
```

```python
mutsig = mutsig[['gene','longname', 'p', 'q']].rename(columns={'gene':'#gene'})
mutsig_RAS = mutsig_RAS[['gene','longname', 'p', 'q']].rename(columns={'gene':'#gene'})
mutsig_nonRAS = mutsig_nonRAS[['gene','longname', 'p', 'q']].rename(columns={'gene':'#gene'})
```

```python
mutsig
```

```python
mutsig.to_csv("../" + data[f'netsig___{dataset}___in'], sep='\t', index=False)
mutsig_RAS.to_csv("../" + data[f'netsig___{dataset}_RAS___in'], sep='\t', index=False)
mutsig_nonRAS.to_csv("../" + data[f'netsig___{dataset}_nonRAS___in'], sep='\t', index=False)
```
