```python
import configparser
import os

import pandas as pd
```

```python
os.chdir('../')
```

```python
dataset = 'all_noHypoHyper'
```

```python
data = configparser.ConfigParser(interpolation=configparser.ExtendedInterpolation())
data.read('../data.ini')
data = data['DEFAULT']
```

### all

```python
dataSamples = pd.read_csv('../' + data[f'{dataset}___samples'], sep='\t', comment='#', engine='python')
dataSamples = pd.DataFrame({'ID': range(0, dataSamples['SAMPLE_ID'].nunique()), 'Sample' : dataSamples['SAMPLE_ID'].unique(), 'Cohort': ['colorectal']*dataSamples['SAMPLE_ID'].nunique()})
dataSamples
```

```python
dataSamples.to_csv('../' + data[f'mutpanning___{dataset}___in_dir'] + '/samples.txt', sep='\t', index=False)
```

```python
dataMutations = pd.read_csv('../' + data[f'{dataset}___mutations'], sep='\t', comment='#', engine='python')
dataMutations
```

```python
dataMutations.to_csv('../' + data[f'mutpanning___{dataset}___in_dir'] + '/mutations.maf', sep='\t', index=False)
```

### RAS and nonRAS

```python
dataSamples_RAS = pd.read_csv('../' + data[f'{dataset}_RAS___samples'], sep='\t', comment='#', engine='python')
dataSamples_RAS = pd.DataFrame({'ID': range(0, dataSamples_RAS['SAMPLE_ID'].nunique()), 'Sample' : dataSamples_RAS['SAMPLE_ID'].unique(), 'Cohort': ['colorectal']*dataSamples_RAS['SAMPLE_ID'].nunique()})

dataSamples_nonRAS = pd.read_csv('../' + data[f'{dataset}_nonRAS___samples'], sep='\t', comment='#', engine='python')
dataSamples_nonRAS = pd.DataFrame({'ID': range(0, dataSamples_nonRAS['SAMPLE_ID'].nunique()), 'Sample' : dataSamples_nonRAS['SAMPLE_ID'].unique(), 'Cohort': ['colorectal']*dataSamples_nonRAS['SAMPLE_ID'].nunique()})
```

```python
dataSamples_RAS.to_csv('../' + data[f'mutpanning___{dataset}_RAS___in_dir'] + '/samples.txt', sep='\t', index=False)

dataSamples_nonRAS.to_csv('../' + data[f'mutpanning___{dataset}_nonRAS___in_dir'] + '/samples.txt', sep='\t', index=False)
```

```python
dataMutations_RAS = pd.read_csv('../' + data[f'{dataset}_RAS___mutations'], sep='\t', comment='#', engine='python')

dataMutations_nonRAS = pd.read_csv('../' + data[f'{dataset}_nonRAS___mutations'], sep='\t', comment='#', engine='python')
```

```python
dataMutations_RAS.to_csv('../' + data[f'mutpanning___{dataset}_RAS___in_dir'] + '/mutations.maf', sep='\t', index=False)

dataMutations_nonRAS.to_csv('../' + data[f'mutpanning___{dataset}_nonRAS___in_dir'] + '/mutations.maf', sep='\t', index=False)
```
