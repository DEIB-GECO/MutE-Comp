```python
import configparser
import os

import pandas as pd
```

```python
os.chdir('../')
```

```python
dataset = 'all_noHypoHyper'
```

```python
data = configparser.ConfigParser(interpolation=configparser.ExtendedInterpolation())
data.read('../data.ini')
data = data['DEFAULT']
```

```python
mutations = pd.read_csv("../" + data[f'{dataset}___mutations'], sep="\t", comment='#')
mutations = mutations[['Hugo_Symbol', 'Tumor_Sample_Barcode']].rename(columns={'Hugo_Symbol': 'GeneSymbol', 'Tumor_Sample_Barcode': 'Sample'})

mutations_RAS = pd.read_csv("../" + data[f'{dataset}_RAS___mutations'], sep="\t", comment='#')
mutations_RAS = mutations_RAS[['Hugo_Symbol', 'Tumor_Sample_Barcode']].rename(columns={'Hugo_Symbol': 'GeneSymbol', 'Tumor_Sample_Barcode': 'Sample'})

mutations_nonRAS = pd.read_csv("../" + data[f'{dataset}_nonRAS___mutations'], sep="\t", comment='#')
mutations_nonRAS = mutations_nonRAS[['Hugo_Symbol', 'Tumor_Sample_Barcode']].rename(columns={'Hugo_Symbol': 'GeneSymbol', 'Tumor_Sample_Barcode': 'Sample'})
```

```python
mutations
```

```python
clinicalSamples = pd.read_csv("../" + data[f'{dataset}___samples'], sep="\t", comment='#')
all_samples = clinicalSamples['SAMPLE_ID'].tolist()
```

```python
pivot = mutations.pivot_table(index='GeneSymbol', columns='Sample', aggfunc='size', fill_value=0)
pivot = pivot.reindex(columns=all_samples, fill_value=0)

pivot_RAS = mutations_RAS.pivot_table(index='GeneSymbol', columns='Sample', aggfunc='size', fill_value=0)
pivot_RAS = pivot_RAS.reindex(columns=all_samples, fill_value=0)

pivot_nonRAS = mutations_nonRAS.pivot_table(index='GeneSymbol', columns='Sample', aggfunc='size', fill_value=0)
pivot_nonRAS = pivot_nonRAS.reindex(columns=all_samples, fill_value=0)
```

```python
pivot
```

```python
pivot.to_csv("../" + data[f'maxmif___{dataset}___in'], sep="\t", index=True)
pivot_RAS.to_csv("../" + data[f'maxmif___{dataset}_RAS___in'], sep="\t", index=True)
pivot_nonRAS.to_csv("../" + data[f'maxmif___{dataset}_nonRAS___in'], sep="\t", index=True)
```
