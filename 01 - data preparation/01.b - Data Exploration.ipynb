## Giannakis

```python
dataMutation_Giannakis = pd.read_csv(data['Giannakis_all___mutations'], sep='\t', comment='#')
```

```python
dataMutation_Giannakis
```

```python
dataMutation_Giannakis_report = dataMutation_Giannakis.profile_report(title='Data Mutations Giannakis Report', explorative=True)
dataMutation_Giannakis_report.to_file(data['profilingReports'] + "Giannakis_mutations_profilingReport.html")
```

```python
# dataMutation_Giannakis_report
```

```python
dataClinical_Giannakis = pd.read_csv(data['Giannakis_all___samples'], sep="\t", comment="#")
```

```python
dataClinical_Giannakis
```

```python
dataClinical_Giannakis_report = dataClinical_Giannakis.profile_report(title='Giannakis Clinical Samples Data Profiling Report')
dataClinical_Giannakis_report.to_file(data['profilingReports'] + "Giannakis_clinicalSamples_profilingReport.html")
```

```python
# dataClinical_Giannakis_report
```

```python
# join dataClinical_Giannakis with dataMutation_Giannakis on SAMPLE_ID = Tumor_Sample_Barcode
merged_data = dataClinical_Giannakis.merge(dataMutation_Giannakis, left_on='SAMPLE_ID', right_on='Tumor_Sample_Barcode')
merged_data
```

### RAS mutated samples

```python
samples_RAS = pd.Series(merged_data[merged_data['Hugo_Symbol'].isin(['HRAS', 'NRAS', 'KRAS'])]['SAMPLE_ID'].unique())

samples_HRAS = pd.Series(merged_data[merged_data['Hugo_Symbol'] == 'HRAS']['SAMPLE_ID'].unique())
samples_NRAS = pd.Series(merged_data[merged_data['Hugo_Symbol'] == 'NRAS']['SAMPLE_ID'].unique())
samples_KRAS = pd.Series(merged_data[merged_data['Hugo_Symbol'] == 'KRAS']['SAMPLE_ID'].unique())

samples_HRAS_NRAS = pd.Series(samples_HRAS[samples_HRAS.isin(samples_NRAS)].unique())
samples_HRAS_KRAS = pd.Series(samples_HRAS[samples_HRAS.isin(samples_KRAS)].unique())
samples_NRAS_KRAS = pd.Series(samples_NRAS[samples_NRAS.isin(samples_KRAS)].unique())

samples_HRAS_NRAS_KRAS = pd.Series(samples_HRAS[samples_HRAS.isin(samples_NRAS_KRAS)].unique())
```

```python
print(f"Number of samples with RAS mutations: {samples_RAS.size}")
print(f"Number of samples with HRAS mutations: {samples_HRAS.size}")
print(f"Number of samples with NRAS mutations: {samples_NRAS.size}")
print(f"Number of samples with KRAS mutations: {samples_KRAS.size}")
print(f"Number of patients with HRAS and NRAS mutations: {samples_HRAS_NRAS.size}")
print(f"Number of patients with HRAS and KRAS mutations: {samples_HRAS_KRAS.size}")
print(f"Number of patients with NRAS and KRAS mutations: {samples_NRAS_KRAS.size}")
print(f"Number of patients with HRAS, NRAS and KRAS mutations: {samples_HRAS_NRAS_KRAS.size}")
```

```python
dataMutation_Giannakis_RAS = dataMutation_Giannakis[dataMutation_Giannakis['Tumor_Sample_Barcode'].isin(samples_RAS)]
dataClinical_Giannakis_RAS = dataClinical_Giannakis[dataClinical_Giannakis['SAMPLE_ID'].isin(samples_RAS)]
dataMutation_Giannakis_nonRAS = dataMutation_Giannakis[~dataMutation_Giannakis['Tumor_Sample_Barcode'].isin(samples_RAS)]
dataClinical_Giannakis_nonRAS = dataClinical_Giannakis[~dataClinical_Giannakis['SAMPLE_ID'].isin(samples_RAS)]
```

```python
dataMutation_Giannakis_RAS.to_csv(data['Giannakis_RAS___mutations'], sep='\t', index=False)
dataClinical_Giannakis_RAS.to_csv(data['Giannakis_RAS___samples'], sep='\t', index=False)
dataMutation_Giannakis_nonRAS.to_csv(data['Giannakis_nonRAS___mutations'], sep='\t', index=False)
dataClinical_Giannakis_nonRAS.to_csv(data['Giannakis_nonRAS___samples'], sep='\t', index=False)
```

```python
del dataMutation_Giannakis, dataMutation_Giannakis_report, dataClinical_Giannakis, dataClinical_Giannakis_report, merged_data, dataMutation_Giannakis_RAS, dataClinical_Giannakis_RAS, dataMutation_Giannakis_nonRAS, dataClinical_Giannakis_nonRAS
```
