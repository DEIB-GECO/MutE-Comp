## Giannakis

### Mutation

```python
dataMutation_Giannakis = pd.read_csv(data['original___Giannakis___mutations'], sep='\t', engine='python', comment='#')
```

```python
dataMutation_Giannakis
```

```python
dataMutation_Giannakis = dataMutation_Giannakis[['Tumor_Sample_Barcode', 'Hugo_Symbol', 'Chromosome', 'Start_Position', 'End_Position', 'Strand', 'Variant_Classification', 'Variant_Type', 'Reference_Allele', 'Tumor_Seq_Allele1', 'Tumor_Seq_Allele2', 'HGVSc', 'HGVSp_Short', 'Protein_position', 'Transcript_ID']]
```

```python
dataMutation_Giannakis
```

```python
uniqueHugo = dataMutation_Giannakis['Hugo_Symbol'].nunique()
uniqueTranscript = dataMutation_Giannakis['Transcript_ID'].nunique()
uniqueHugo, uniqueTranscript
```

```python
unknownHugo = (dataMutation_Giannakis['Hugo_Symbol'] == 'Unknown').sum()
emptyHugo = (dataMutation_Giannakis['Hugo_Symbol'] == '').sum()
naHugo = pd.isna(dataMutation_Giannakis['Hugo_Symbol']).sum()

unknownTranscript = (dataMutation_Giannakis['Transcript_ID'] == 'Unknown').sum()
emptyTranscript = (dataMutation_Giannakis['Transcript_ID'] == '').sum()
naTranscript = pd.isna(dataMutation_Giannakis['Transcript_ID']).sum()

unknownHugoUnknownTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == 'Unknown']['Transcript_ID'] == 'Unknown').sum()
emptyHugoUnknownTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == '']['Transcript_ID'] == 'Unknown').sum()
naHugoUnknownTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'].isna()]['Transcript_ID'] == 'Unknown').sum()
unknownHugoEmptyTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == 'Unknown']['Transcript_ID'] == '').sum()
emptyHugoEmptyTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == '']['Transcript_ID'] == '').sum()
naHugoEmptyTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'].isna()]['Transcript_ID'] == '').sum()
unknownHugoNaTranscript = (pd.isna(dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == 'Unknown']['Transcript_ID'])).sum()
emptyHugoNaTranscript = (pd.isna(dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == '']['Transcript_ID'])).sum()
naHugoNaTranscript = (pd.isna(dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'].isna()]['Transcript_ID'])).sum()

unknownHugoEmptyTranscript = (dataMutation_Giannakis[dataMutation_Giannakis['Hugo_Symbol'] == 'Unknown']['Transcript_ID'] == '').sum()

(unknownHugo, emptyHugo, naHugo), (unknownTranscript, emptyTranscript, naTranscript), ((unknownHugoUnknownTranscript, emptyHugoUnknownTranscript, naHugoUnknownTranscript), (unknownHugoEmptyTranscript, emptyHugoEmptyTranscript, naHugoEmptyTranscript), (unknownHugoNaTranscript, emptyHugoNaTranscript, naHugoNaTranscript))
```

```python
# remove rows with both Hugo_Symbol == 'Unknown' and Transcript_ID == NA
dataMutation_Giannakis_nona = dataMutation_Giannakis[~((dataMutation_Giannakis['Hugo_Symbol'] == 'Unknown') & pd.isna(dataMutation_Giannakis['Transcript_ID']))]
dataMutation_Giannakis_nona
```

```python
# Get unique transcript ids
uniqueTranscriptIds = dataMutation_Giannakis_nona['Transcript_ID'].unique()

# Create a dictionary mapping from Transcript_ID to Hugo_Symbol
transcriptIds_HugoSymbol_mapping = {id: ensembl.gene_name_of_transcript_id(id) for id in uniqueTranscriptIds}

transcriptIds_HugoSymbol_mapping
```

```python
# add Ensembl Hugo_Symbol
dataMutation_Giannakis_cleaned = dataMutation_Giannakis_nona.copy()
dataMutation_Giannakis_cleaned.rename(columns={'Hugo_Symbol': 'original_Hugo_Symbol'}, inplace=True)
dataMutation_Giannakis_cleaned.insert(1, 'Hugo_Symbol', '')
dataMutation_Giannakis_cleaned.loc[:,'Hugo_Symbol'] = dataMutation_Giannakis_cleaned['Transcript_ID'].map(transcriptIds_HugoSymbol_mapping)
```

```python
dataMutation_Giannakis_cleaned
```

```python
# show rows where Hugo_Symbol and original_Hugo_Symbol are different
dataMutation_Giannakis_cleaned.loc[dataMutation_Giannakis_cleaned['Hugo_Symbol'] != dataMutation_Giannakis_cleaned['original_Hugo_Symbol'], ['Transcript_ID', 'Hugo_Symbol', 'original_Hugo_Symbol']]
```

```python
# add gene length
dataMutation_Giannakis_cleaned['Gene_Length'] = dataMutation_Giannakis_cleaned['Hugo_Symbol'].apply(lambda x: ensembl.genes_by_name(x)[0].length)
```

```python
dataMutation_Giannakis_cleaned
```

```python
uniqueHugo = dataMutation_Giannakis_cleaned['Hugo_Symbol'].nunique(dropna=False)
uniqueTranscript = dataMutation_Giannakis_cleaned['Transcript_ID'].nunique(dropna=False)
uniqueHugo, uniqueTranscript
```

```python
unknownHugo = (dataMutation_Giannakis_cleaned['Hugo_Symbol'] == 'Unknown').sum()
emptyHugo = (dataMutation_Giannakis_cleaned['Hugo_Symbol'] == '').sum()
naHugo = pd.isna(dataMutation_Giannakis_cleaned['Hugo_Symbol']).sum()

unknownTranscript = (dataMutation_Giannakis_cleaned['Transcript_ID'] == 'Unknown').sum()
emptyTranscript = (dataMutation_Giannakis_cleaned['Transcript_ID'] == '').sum()
naTranscript = pd.isna(dataMutation_Giannakis_cleaned['Transcript_ID']).sum()

unknownHugoUnknownTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == 'Unknown']['Transcript_ID'] == 'Unknown').sum()
emptyHugoUnknownTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == '']['Transcript_ID'] == 'Unknown').sum()
naHugoUnknownTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'].isna()]['Transcript_ID'] == 'Unknown').sum()
unknownHugoEmptyTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == 'Unknown']['Transcript_ID'] == '').sum()
emptyHugoEmptyTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == '']['Transcript_ID'] == '').sum()
naHugoEmptyTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'].isna()]['Transcript_ID'] == '').sum()
unknownHugoNaTranscript = (pd.isna(dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == 'Unknown']['Transcript_ID'])).sum()
emptyHugoNaTranscript = (pd.isna(dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == '']['Transcript_ID'])).sum()
naHugoNaTranscript = (pd.isna(dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'].isna()]['Transcript_ID'])).sum()

unknownHugoEmptyTranscript = (dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Hugo_Symbol'] == 'Unknown']['Transcript_ID'] == '').sum()

(unknownHugo, emptyHugo, naHugo), (unknownTranscript, emptyTranscript, naTranscript), ((unknownHugoUnknownTranscript, emptyHugoUnknownTranscript, naHugoUnknownTranscript), (unknownHugoEmptyTranscript, emptyHugoEmptyTranscript, naHugoEmptyTranscript), (unknownHugoNaTranscript, emptyHugoNaTranscript, naHugoNaTranscript))
```

```python
dataMutation_Giannakis_cleaned['Mutation_Length'] = dataMutation_Giannakis_cleaned['End_Position'] - dataMutation_Giannakis_cleaned['Start_Position'] + 1
dataMutation_Giannakis_cleaned
```

```python
### TODO: what to do with missing protein positions? For now, removing them
dataMutation_Giannakis_cleaned = dataMutation_Giannakis_cleaned[~dataMutation_Giannakis_cleaned['Protein_position'].isna()].copy()
dataMutation_Giannakis_cleaned
```

```python
### TODO: Is this correct?
transcriptIds_proteinIds_mapping = {tid: (len(ensembl.transcript_by_id(tid).protein_sequence) if ensembl.transcript_by_id(tid).protein_sequence is not None else np.nan) for tid in dataMutation_Giannakis_cleaned['Transcript_ID'].unique()}
transcriptIds_proteinIds_mapping
```

```python
dataMutation_Giannakis_cleaned['Protein_Length'] = dataMutation_Giannakis_cleaned['Transcript_ID'].map(transcriptIds_proteinIds_mapping)
dataMutation_Giannakis_cleaned
```

```python
dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Protein_Length'].isna()]
```

```python
### TODO: What to do with nan protein length? for now remove
dataMutation_Giannakis_cleaned = dataMutation_Giannakis_cleaned[~dataMutation_Giannakis_cleaned['Protein_Length'].isna()]
```

```python
dataMutation_Giannakis_cleaned[dataMutation_Giannakis_cleaned['Protein_Length'].isna()]
```

```python
#save cleaned data
dataMutation_Giannakis_cleaned.to_csv(data['Giannakis_all___mutations'], sep="\t", index=False)
```

```python
# clean memory
del dataMutation_Giannakis, dataMutation_Giannakis_nona, dataMutation_Giannakis_cleaned
```
